### Import Dependencies

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
```

### Get File Paths

```python
BASE_DIR = Path.cwd().parent
PLANET_PATH = BASE_DIR / 'data' / 'raw' / 'raw_planet_data.csv'
STAR_PATH = BASE_DIR / 'data' / 'raw' / 'raw_star_data.csv'
```

### Load Data

```python
planet_data = pd.read_csv(PLANET_PATH)
star_data = pd.read_csv(STAR_PATH)
```

### Drop Eccentricity

```python
filtered_planet_data = planet_data.drop(columns=['eccentricity'])
```

```python
complete_planet_data = filtered_planet_data[filtered_planet_data.notna().all(axis=1)]
complete_planet_data
```

### Filter Star Data

```python
filtered_star_data = star_data[star_data['star_name'].isin(complete_planet_data['star_name'])]
```

```python
filtered_star_data = filtered_star_data.assign(non_null_count=star_data.notna().sum(axis=1)) \
              .sort_values(by='non_null_count', ascending=False) \
              .drop(columns='non_null_count')

filtered_star_data = filtered_star_data.drop_duplicates(subset='star_name', keep='first')
filtered_star_data
```

```python
filtered_star_data.isna().sum()
```

### Drop Columns with Lots of Null

```python
filtered_star_data = filtered_star_data.drop(columns = ['star_type','star_age','star_metallicity'])
```

```python
filtered_star_data.notna().all(axis=1).sum()
```

### Complete Datasets

```python
complete_star_data = filtered_star_data[filtered_star_data.notna().all(axis=1)]
complete_star_data
```

```python
complete_planet_data = complete_planet_data[complete_planet_data['star_name'].isin(complete_star_data['star_name'])]
```

```python
complete_planet_data
```
